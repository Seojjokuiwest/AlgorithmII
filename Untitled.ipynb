```python
class MyGRUCell(nn.Module):
    def __init__(self, d, h):
        super().__init__()

        # reset r, update z를 한 번에 계산
        self.W_x = nn.Linear(d, 2 * h)
        self.W_h = nn.Linear(h, 2 * h, bias=False)

        # candidate hidden state
        self.W_xh = nn.Linear(d, h)
        self.W_hh = nn.Linear(h, h, bias=False)

    def forward(self, x, state, return_gates=False):
        h, _ = state
        
class TorchGRUCell(nn.Module):
    def __init__(self, d, h):
        super().__init__()
        self.cell = nn.GRUCell(d, h)

    def forward(self, x, state):
        h, _ = state
        h = self.cell(x, h)

        return h, (h, None)
        
class CharLM(nn.Module):
    def __init__(self, V, d=64, h=128, cell='lstm'):
        super().__init__()
        self.h = h
        self.emb = nn.Embedding(V, d)

        if cell == 'lstm':
            self.cell = MyLSTMCell(d, h)
        elif cell == 'gru':
            self.cell = MyGRUCell(d, h)
        elif cell == 'gru_torch':
            self.cell = TorchGRUCell(d, h)
        else:
            self.cell = MyRNNCell(d, h)

        self.head = nn.Linear(h, V)

    def init_state(self, B):
        z = torch.zeros(B, self.h, device=device)
        return (z, z.clone())

    def forward(self, x, state=None, collect_gates=False):
        B, L = x.shape
        state = self.init_state(B) if state is None else state
        e = self.emb(x)

        outs, gate_log = [], []

        for t in range(L):
            if collect_gates:
                hh, state, gts = self.cell(
                    e[:, t],
                    state,
                    return_gates=True
                )
                gate_log.append(gts)
            else:
                hh, state = self.cell(e[:, t], state)

            outs.append(hh)

        logits = self.head(torch.stack(outs, 1))

        return (
            (logits, state, gate_log)
            if collect_gates
            else (logits, state)
        )
```

```python
def train(model, steps=900, batch_size=64, seq_len=48,
          lr=3e-3, log_every=150):

    model.to(device).train()

    opt = torch.optim.Adam(
        model.parameters(),
        lr=lr
    )

    # 두 모델이 동일한 학습 배치를 사용하도록 고정
    g = torch.Generator().manual_seed(SEED)

    hist = []
    t0 = time.time()

    for s in range(1, steps + 1):

        x, y = get_batch(
            train_ids,
            batch_size,
            seq_len,
            generator=g
        )

        logits, _ = model(x)

        loss = F.cross_entropy(
            logits.reshape(-1, V),
            y.reshape(-1)
        )

        opt.zero_grad()
        loss.backward()

        nn.utils.clip_grad_norm_(
            model.parameters(),
            1.0
        )

        opt.step()

        hist.append(loss.item())

        if s % log_every == 0:
            print(
                f'  step {s:4d}/{steps}  '
                f'loss {np.mean(hist[-50:]):.4f}'
            )

    elapsed = time.time() - t0
    print(f'  학습 시간 {elapsed:.1f}초')

    return hist


@torch.no_grad()
def evaluate(model, ids,
             batch_size=128,
             seq_len=48,
             n_batch=20):

    model.eval()

    # 두 모델이 동일한 평가 배치를 사용하도록 고정
    g = torch.Generator().manual_seed(1234)

    losses = []
    accs = []

    for _ in range(n_batch):

        x, y = get_batch(
            ids,
            batch_size,
            seq_len,
            generator=g
        )

        logits, _ = model(x)

        loss = F.cross_entropy(
            logits.reshape(-1, V),
            y.reshape(-1)
        )

        acc = (
            logits.argmax(-1) == y
        ).float().mean()

        losses.append(loss.item())
        accs.append(acc.item())

    return (
        float(np.mean(losses)),
        float(np.mean(accs))
    )


# ---------------------------------------------------------
# 두 GRU 모델을 동일한 기본 설정으로 다시 생성
# ---------------------------------------------------------

torch.manual_seed(SEED)
gru_manual = CharLM(
    V,
    d=64,
    h=128,
    cell='gru'
).to(device)

torch.manual_seed(SEED)
gru_torch = CharLM(
    V,
    d=64,
    h=128,
    cell='gru_torch'
).to(device)


# ---------------------------------------------------------
# 학습
# ---------------------------------------------------------

print('[모델 A] 직접 구현 GRU')
hist_manual = train(
    gru_manual,
    steps=900,
    batch_size=64,
    seq_len=48,
    lr=3e-3
)

print()

print('[모델 B] nn.GRUCell')
hist_torch = train(
    gru_torch,
    steps=900,
    batch_size=64,
    seq_len=48,
    lr=3e-3
)


# ---------------------------------------------------------
# 학습/시험 성능 평가
# ---------------------------------------------------------

manual_train_loss, manual_train_acc = evaluate(
    gru_manual,
    train_ids
)

manual_test_loss, manual_test_acc = evaluate(
    gru_manual,
    test_ids
)

torch_train_loss, torch_train_acc = evaluate(
    gru_torch,
    train_ids
)

torch_test_loss, torch_test_acc = evaluate(
    gru_torch,
    test_ids
)


# 파라미터 수
def count_params(model):
    return sum(
        p.numel()
        for p in model.parameters()
        if p.requires_grad
    )


print()
print(
    f"{'모델':<22}"
    f"{'파라미터':>12}"
    f"{'학습 loss':>12}"
    f"{'학습 acc':>12}"
    f"{'시험 loss':>12}"
    f"{'시험 acc':>12}"
)

print('-' * 82)

print(
    f"{'직접 구현 GRU':<22}"
    f"{count_params(gru_manual):>12,}"
    f"{manual_train_loss:>12.4f}"
    f"{manual_train_acc:>12.4f}"
    f"{manual_test_loss:>12.4f}"
    f"{manual_test_acc:>12.4f}"
)

print(
    f"{'nn.GRUCell':<22}"
    f"{count_params(gru_torch):>12,}"
    f"{torch_train_loss:>12.4f}"
    f"{torch_train_acc:>12.4f}"
    f"{torch_test_loss:>12.4f}"
    f"{torch_test_acc:>12.4f}"
)
```
